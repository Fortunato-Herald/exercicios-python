```python
"""
Teste Grafico
"""

# Importa a biblioteca matplotlib para criação de gráficos
import matplotlib.pyplot as plt

# Lista com os nomes (rótulos) de cada fatia do gráfico
labels = ['A', 'B', 'C', 'D', 'E', 'F', 'G']

# Lista com os valores correspondentes a cada fatia
valores = [10, 20, 15, 25, 30, 40, 50]

# Lista de cores personalizadas para cada fatia do gráfico
cores = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8', '#C7CEEA', '#FFDAC1']

# Lista que define quais fatias terão destaque (explode)
# 0 = normal | valor > 0 = fatia "afastada" do centro
explode = [0, 0.05, 0, 0, 0.08, 0, 0]

# Define o tamanho da figura (largura, altura em polegadas)
plt.figure(figsize=(8, 8))

# Cria o gráfico de pizza
plt.pie(
    valores,                # Dados numéricos (tamanho das fatias)
    labels=labels,          # Rótulos das fatias
    colors=cores,           # Cores personalizadas
    autopct='%6.1f%%',      # Mostra porcentagem com 1 casa decimal
    startangle=190,         # Rotaciona o gráfico para melhor visualização
    explode=explode,        # Destaca algumas fatias
    shadow=True             # Adiciona sombra ao gráfico
)

# Define o título do gráfico
plt.title("Gráfico de Pizza Estilizado 🔥")

# Garante que o gráfico fique em formato de círculo (não oval)
plt.axis('equal')

# Exibe o gráfico na tela
plt.show()
```

```python
"""
1. Leia um número e mostre antecessor e sucessor
"""
n = int(input('Numero: '))

ant = n - 1
suc = n + 1

print(f'Antecessor: {ant}')
print(f'Sucessor: {suc}')
```

```python
"""
2. Leia um número e mostre dobro, triplo e raiz
"""
import math

n = int(input('Numero: '))

d = n * 2
t = n * 3
r = math.sqrt(n)

print(f'Dobro: {d}')
print(f'Triplo: {t}')
print(f'Raiz: {r}')


```

```python
"""
3. Leia 2 números e mostre a soma
"""

n = int(input('Numero 1: '))
n2 = int(input('Numero 2: '))

s = n + n2

print(f'Soma: {s}')
```

```python
"""
4. Leia 2 números e mostre média
"""

n1 = float(input('Numero 1: '))
n2 = float(input('Numero 2: '))

m = (n1 + n2) / 2

print(f'Media: {m}')
```

```python
"""
5. Leia metros e converta para cm e mm
"""

m = float(input('Informe o valor em METROS: '))

cm = m * 100
mm = m * 1000

print(f'Valor convertido em centimetros: {cm} CM')
print(f'Valor convertido em milimetros: {mm} MM')
```

```python
"""
6. Leia um valor e aplique 10% de desconto
"""
preco = float(input('Valor do produto: R$ '))

desconto = preco * 0.10
preco_final = preco - desconto

print(f'Valor do desconto: R$ {desconto:.2f} R$')
print(f'Valor do produto com desconto: R$ {preco_final:.2f} R$')

```

```python
"""
7. Leia salário e aumente 15%
"""

salario = float(input('Salário atual: R$ '))

sal = salario
aumento = salario * 0.15
salario_final = salario + aumento

print(f'Salário atual: R$ {sal:.2f} R$')
print(f'Valor do aumento: R$ {aumento:.2f} R$')
print(f'Valor do salário com aumento: R$ {salario_final:.2f} R$')

```

```python
"""
8. Leia temperatura em °C e converta para °F
"""

c = float(input('Informe a temperatura em °C: '))

f = (c * 9/5) + 32

print(f'Temperatura convertida: {f:.2f} °F')
```
